```python
import pandas as pd

colaboradores = pd.read_excel("../datos/DATASET_BASE_OTC.xlsx", sheet_name="Colaboradores")
df = pd.DataFrame(colaboradores)

def crearcolumnas():
    df["SalarioDiario"] = df["Sueldo Neto Total"] / 160
    df["CostoInterno"] = df["Sueldo Neto IMSS"] + df["Sueldo Complementos"]
    return

def Limpieza():
    df.drop(df.columns[0], axis=1, inplace=True)
    return

def estadisticasreferencia():
    SalarioPromedio = df["Sueldo Neto Total"].mean()
    SueldoPromedio = df["Sueldo Neto IMSS"].mean()
    CompensacionPromedio = df["Sueldo Complementos"].mean()
    CostoInternoPromedio = df["CostoInterno"].mean()
    
    return

def actualiza():
    # index=False evita que se guarde una columna extra con los números de fila
    df.to_excel("../datos/archivocambiado.xlsx", index=False)
    return

crearcolumnas()
Limpieza()
estadisticasreferencia()
actualiza()
```
